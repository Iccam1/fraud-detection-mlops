```python
import matplotlib.pyplot as plt

# Plot TransactionAmt distribution
plt.figure(figsize=(10, 5))
plt.hist(df["TransactionAmt"], bins=100, edgecolor='black')
plt.xlabel("Transaction Amount ($)")
plt.ylabel("Frequency")
plt.title("Transaction Amount Distribution")
plt.show()
```

```python
# Better view with log scale
plt.figure(figsize=(10, 5))
plt.hist(df["TransactionAmt"], bins=100, edgecolor='black')
plt.xlabel("Transaction Amount ($)")
plt.ylabel("Frequency")
plt.title("Transaction Amount Distribution (Log Scale)")
plt.yscale('log')
plt.show()
```

```python
# Fraud vs Legitimate amounts
plt.figure(figsize=(10, 5))
df[df["isFraud"]==0]["TransactionAmt"].hist(bins=100, alpha=0.5, label="Legitimate", edgecolor='black')
df[df["isFraud"]==1]["TransactionAmt"].hist(bins=100, alpha=0.5, label="Fraud", edgecolor='black')
plt.xlabel("Transaction Amount ($)")
plt.ylabel("Frequency")
plt.legend()
plt.title("Transaction Amount: Fraud vs Legitimate")
plt.show()
```

```python
# Side by side comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Legitimate
axes[0].hist(df[df["isFraud"]==0]["TransactionAmt"], bins=100, edgecolor='black')
axes[0].set_title("Legitimate Transactions")
axes[0].set_xlabel("Amount ($)")
axes[0].set_ylabel("Frequency")

# Fraud
axes[1].hist(df[df["isFraud"]==1]["TransactionAmt"], bins=100, edgecolor='black', color='orange')
axes[1].set_title("Fraudulent Transactions")
axes[1].set_xlabel("Amount ($)")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# Compare average amounts
print("Legitimate avg:", df[df["isFraud"]==0]["TransactionAmt"].mean())
print("Fraud avg:", df[df["isFraud"]==1]["TransactionAmt"].mean())
```

```python
# ProductCD fraud rate
df.groupby("ProductCD")["isFraud"].agg(['count', 'sum', 'mean']).sort_values('mean', ascending=False)
```

```python
# Card type fraud rate
df.groupby("card4")["isFraud"].agg(['count', 'sum', 'mean']).sort_values('mean', ascending=False)
```

```python
# Email domain fraud rates (purchaser email)
df.groupby("P_emaildomain")["isFraud"].agg(['count', 'sum', 'mean']).sort_values('mean', ascending=False).head(10)
```

```python
# Recipient email fraud rates
df.groupby("R_emaildomain")["isFraud"].agg(['count', 'sum', 'mean']).sort_values('mean', ascending=False).head(10)
```
